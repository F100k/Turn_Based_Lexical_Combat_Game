```python
import wikipediaapi

# 1. กำหนดหัวข้อหลักในแต่ละหมวด (เป้าหมาย 1,500+ ประโยคต่อหมวด)
categories = {
    "Business": [
        "Business", "Corporate finance", "Management", "Marketing", "Enterprise",
        "Venture capital", "Economics", "Stock market", "Entrepreneurship", "Supply chain",
        "Investment", "Accounting", "Microeconomics", "Macroeconomics", "E-commerce"
    ],
    "Technology": [
        "Computer science", "Artificial intelligence", "Software engineering", "Algorithm",
        "Information technology", "Cybersecurity", "Cloud computing", "Data science",
        "Microprocessor", "Computer network", "Operating system", "Database",
        "Programming language", "Machine learning", "Web development"
    ],
    "Science": [
        "Physics", "Chemistry", "Biology", "Quantum mechanics", "Scientific method",
        "Ecology", "Astronomy", "Genetics", "Thermodynamics", "Neuroscience",
        "Evolution", "Organic chemistry", "Astrophysics", "Cell biology", "Molecular biology"
    ],
    "Arts": [
        "Art history", "Painting", "Music theory", "Literature", "Architecture",
        "Sculpture", "Classical music", "Theatre", "Poetry", "Philosophy of art",
        "Fine art", "Modern art", "Literary criticism", "Symphony", "Renaissance art"
    ],
    "Fantasy": [
        "Fantasy", "Magic (paranormal)", "Mythology", "Dragon", "Fantasy literature",
        "Folklore", "Legend", "Fairytale", "High fantasy", "Supernatural",
        "Urban fantasy", "Dark fantasy", "Magic in fiction", "Mythological creature", "Wizard",
        "Ghost", "Vampire", "Elf (mythology)", "Dwarf (mythology)", "Sorcery"
    ],
    "History": [
        "World history", "Ancient history", "Middle Ages", "Industrial Revolution",
        "Historiography", "Roman Empire", "World War I",
        "Renaissance", "Ancient Egypt"
    ]
}

# 2. ตั้งค่า Wikipedia API
wiki = wikipediaapi.Wikipedia(
    user_agent='TextGameProject-ProMax/1.0 (Academic Project; kmutt.ac.th)',
    language='en'
)

output_filename = "corpus.txt"
total_sentences = 0

print("=== เริ่มต้นกระบวนการสร้าง Corpus ขนาดใหญ่ ===")
with open(output_filename, "w", encoding="utf-8") as f:
    for domain, topics in categories.items():
        print(f"📥 กำลังดึงข้อมูลเชิงลึกหมวด: [{domain}]...")
        domain_count = 0

        for topic in topics:
            page = wiki.page(topic)
            if page.exists():
                sentences = page.text.split('\n')
                for sentence in sentences:
                    clean_sentence = sentence.strip().lower()
                    # กรองเงื่อนไขความยาวประโยค
                    if len(clean_sentence) > 25 and not clean_sentence.startswith('==='):
                        f.write(clean_sentence + "\n")
                        domain_count += 1
                        total_sentences += 1
            else:
                print(f"  - ข้ามหัวข้อที่ไม่พบ: {topic}")

        print(f"  -> หมวด {domain} สำเร็จ: ได้เพิ่มมา {domain_count} ประโยค\n")

print(f"🎉 สำเร็จเรียบร้อย! corpus.txt เวอร์ชันสมดุลมีข้อมูลรวมทั้งหมด {total_sentences} ประโยค พร้อมนำไปเทรนโมเดลแล้วครับ")
```

```python
print(f"--- ตัวอย่าง 10 บรรทัดแรกใน {output_filename} ---")
with open("corpus.txt", "r", encoding="utf-8") as f:
    for i in range(10):
        line = f.readline()
        if line:
            print(f"{i+1}: {line.strip()}")
        else:
            break
```
